```python
from pathlib import Path

import pandas as pd

from utils.custom_process_structure import (
    run_composition_batch,
    run_drift_batch,
    run_simple_batch,
)
GLOBAL_RANDOM_SEED = 123
x_max_in_graphs = 1e4
y_max_in_graphs = 1e4
RESULTS_ROOT = Path("results") / "experiments"
POWER_LAW_SINGLE_XOR = {"m": 1, "n": 1, "o": 100, "p": 0.0, "q": "power_law"}

PRESET_EXPERIMENTS = [
    {
        "id": "example1",
        "name": "Example configuration",
        "experiment_type": "example",
        "output_folder": RESULTS_ROOT / "examples",
        "process_configuration": {
            "N": 10000,
            "m": 2,
            "n": 2,
            "o": 3,
            "p": 0.2,
            "q": "majority",
            "alpha": None,
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "a1",
        "name": "Single equal XOR",
        "experiment_type": "simple",
        "output_folder": RESULTS_ROOT / "simple",
        "process_configuration": {
            "N": 10000,
            "m": 1,
            "n": 1,
            "o": 10,
            "p": 0.0,
            "q": "equal",
            "alpha": None,
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "a2",
        "name": "Single majority XOR",
        "experiment_type": "simple",
        "output_folder": RESULTS_ROOT / "simple",
        "process_configuration": {
            "N": 10000,
            "m": 1,
            "n": 1,
            "o": 10,
            "p": 0.0,
            "q": "majority",
            "alpha": None,
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "a3",
        "name": "Single power law XOR",
        "experiment_type": "simple",
        "output_folder": RESULTS_ROOT / "simple",
        "process_configuration": {
            "N": 10000,
            "m": 1,
            "n": 1,
            "o": 10,
            "p": 0.0,
            "q": "power_law",
            "alpha": 2,
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "a4",
        "name": "Multi-equal XORs",
        "experiment_type": "simple",
        "output_folder": RESULTS_ROOT / "simple",
        "process_configuration": {
            "N": 10000,
            "m": 1,
            "n": 6,
            "o": 10,
            "p": 0.0,
            "q": "equal",
            "alpha": None,
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "a5",
        "name": "Multi-majority XORs",
        "experiment_type": "simple",
        "output_folder": RESULTS_ROOT / "simple",
        "process_configuration": {
            "N": 10000,
            "m": 1,
            "n": 6,
            "o": 10,
            "p": 0.0,
            "q": "majority",
            "alpha": None,
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "a6",
        "name": "Multi-power law XORs",
        "experiment_type": "simple",
        "output_folder": RESULTS_ROOT / "simple",
        "process_configuration": {
            "N": 10000,
            "m": 1,
            "n": 6,
            "o": 10,
            "p": 0.0,
            "q": "power_law",
            "alpha": 2,
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "a7",
        "name": "Single majority XOR, loop",
        "experiment_type": "simple",
        "output_folder": RESULTS_ROOT / "simple",
        "process_configuration": {
            "N": 10000,
            "m": 1,
            "n": 1,
            "o": 10,
            "p": 0.2,
            "q": "majority",
            "alpha": None,
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "a8",
        "name": "Single power law XOR, loop",
        "experiment_type": "simple",
        "output_folder": RESULTS_ROOT / "simple",
        "process_configuration": {
            "N": 10000,
            "m": 1,
            "n": 1,
            "o": 10,
            "p": 0.2,
            "q": "power_law",
            "alpha": 2,
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "a9",
        "name": "Parallel, majority XOR",
        "experiment_type": "simple",
        "output_folder": RESULTS_ROOT / "simple",
        "process_configuration": {
            "N": 10000,
            "m": 2,
            "n": 1,
            "o": 10,
            "p": 0.0,
            "q": "majority",
            "alpha": None,
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "a10",
        "name": "Parallel, power law XOR",
        "experiment_type": "simple",
        "output_folder": RESULTS_ROOT / "simple",
        "process_configuration": {
            "N": 10000,
            "m": 2,
            "n": 1,
            "o": 10,
            "p": 0.0,
            "q": "power_law",
            "alpha": 2,
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "a11",
        "name": "Complex majority XORs",
        "experiment_type": "simple",
        "output_folder": RESULTS_ROOT / "simple",
        "process_configuration": {
            "N": 10000,
            "m": 2,
            "n": 6,
            "o": 10,
            "p": 0.2,
            "q": "majority",
            "alpha": None,
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "b1",
        "name": "Composition, $\\alpha$ constant",
        "experiment_type": "composition",
        "output_folder": RESULTS_ROOT / "composition",
        "process_configuration": {
            "N": 10000,
            "a": {**POWER_LAW_SINGLE_XOR, "alpha": 1.0},
            "b": {**POWER_LAW_SINGLE_XOR, "alpha": 1.0},
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "b2",
        "name": "Composition, $\\alpha$ change",
        "experiment_type": "composition",
        "output_folder": RESULTS_ROOT / "composition",
        "process_configuration": {
            "N": 10000,
            "a": {**POWER_LAW_SINGLE_XOR, "alpha": 1.0},
            "b": {**POWER_LAW_SINGLE_XOR, "alpha": 3.0},
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "c1",
        "name": "Non-overlapping change",
        "experiment_type": "drift",
        "output_folder": RESULTS_ROOT / "drift",
        "process_configuration": {
            "N": 10000,
            "s": "non_overlapping",
            "a": {**POWER_LAW_SINGLE_XOR, "alpha": 2.0},
            "b": {**POWER_LAW_SINGLE_XOR, "alpha": 2.0},
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "c2",
        "name": "$\\alpha$ change, overlapping",
        "experiment_type": "drift",
        "output_folder": RESULTS_ROOT / "drift",
        "process_configuration": {
            "N": 10000,
            "s": "overlapping",
            "a": {**POWER_LAW_SINGLE_XOR, "alpha": 1.0},
            "b": {**POWER_LAW_SINGLE_XOR, "alpha": 3.0},
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
    {
        "id": "c3",
        "name": "XOR change, overlapping",
        "experiment_type": "drift",
        "output_folder": RESULTS_ROOT / "drift",
        "process_configuration": {
            "N": 10000,
            "s": "overlapping",
            "a": {"m": 1, "n": 2, "o": 10, "p": 0.0, "q": "majority", "alpha": None},
            "b": {"m": 1, "n": 3, "o": 10, "p": 0.0, "q": "majority", "alpha": None},
            "seed": GLOBAL_RANDOM_SEED,
        },
    },
]

EXPERIMENT_TYPE_LABELS = {
    "example": "Example",
    "simple": "Simple process structure",
    "composition": "Process composition",
    "drift": "Process change",
}


def to_runner_settings(experiment: dict) -> dict:
    return {"name": experiment["id"], **experiment["process_configuration"]}


def build_overview_table(experiments: list[dict]):
    rows = [
        {
            "Scenario": EXPERIMENT_TYPE_LABELS[experiment["experiment_type"]],
            "Experiment": f"{experiment['id']}: {experiment['name']}",
        }
        for experiment in experiments
    ]
    return pd.DataFrame(rows)


PROCESS_PARAM_KEYS = ("m", "n", "o", "p", "q", "alpha")


def _split_process_configuration(config: dict) -> tuple[dict, dict, str | None, int | None, int | None]:
    if "a" in config:
        return (
            config["a"],
            config.get("b", {}),
            config.get("s"),
            config.get("N"),
            config.get("seed"),
        )
    process_a = {key: config.get(key) for key in PROCESS_PARAM_KEYS}
    return process_a, {}, None, config.get("N"), config.get("seed")


def build_parameter_overview_table(experiments: list[dict]) -> pd.DataFrame:
    columns = pd.MultiIndex.from_tuples(
        [
            ("Scenario", ""),
            ("ID", ""),
            ("Name", ""),
            ("N", ""),
            *[("Process a", key) for key in PROCESS_PARAM_KEYS],
            ("s", ""),
            *[("Process b", key) for key in PROCESS_PARAM_KEYS],
            ("seed", ""),
        ]
    )
    rows = []
    for experiment in experiments:
        process_a, process_b, s, n, seed = _split_process_configuration(
            experiment["process_configuration"]
        )
        rows.append(
            {
                ("Scenario", ""): EXPERIMENT_TYPE_LABELS[experiment["experiment_type"]],
                ("ID", ""): experiment["id"],
                ("Name", ""): experiment["name"],
                ("N", ""): n,
                **{("Process a", key): process_a.get(key) for key in PROCESS_PARAM_KEYS},
                ("s", ""): s,
                **{
                    ("Process b", key): process_b.get(key) if process_b else None
                    for key in PROCESS_PARAM_KEYS
                },
                ("seed", ""): seed,
            }
        )
    return pd.DataFrame(rows, columns=columns)


def group_experiments_for_batch(experiments: list[dict]) -> dict[tuple[str, Path], list[dict]]:
    groups: dict[tuple[str, Path], list[dict]] = {}
    for experiment in experiments:
        key = (experiment["experiment_type"], experiment["output_folder"])
        groups.setdefault(key, []).append(to_runner_settings(experiment))
    return groups


def run_preset_experiments(
    experiments: list[dict],
    trace_count: int = 15,
    x_max_in_graphs: float | None = None,
    y_max_in_graphs: float | None = None,
) -> list[dict]:
    runners = {
        "example": run_simple_batch,
        "simple": run_simple_batch,
        "composition": run_composition_batch,
        "drift": run_drift_batch,
    }
    all_outputs: list[dict] = []

    for (experiment_type, output_folder), batch in group_experiments_for_batch(experiments).items():
        scenario = EXPERIMENT_TYPE_LABELS[experiment_type]
        print(f"Running {scenario} experiments in {output_folder} ...")
        batch_outputs = runners[experiment_type](
            batch,
            results_dir=output_folder,
            trace_count=trace_count,
            x_max_in_graphs=x_max_in_graphs,
            y_max_in_graphs=y_max_in_graphs,
        )
        for item in batch_outputs:
            print(f"Running {item['name']} ...")
            if experiment_type in {"example", "simple"}:
                print(f"Saved RFC PDF: {item['rfc_pdf']}")
                print(f"Saved BPMN PDF: {item['bpmn_pdf']}")
            else:
                print(f"Saved BPMN a: {item['bpmn_pdf_a']}")
                print(f"Saved BPMN b: {item['bpmn_pdf_b']}")
                print(f"Saved RFC a: {item['rfc_pdf_a']}")
                print(f"Saved RFC b: {item['rfc_pdf_b']}")
                print(f"Saved RFC combined: {item['rfc_pdf_combined']}")
            print()
            all_outputs.append(item)

    return all_outputs
```
